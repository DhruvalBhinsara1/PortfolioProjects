```python
class Bank:

    def __init__(self, salary, name,  account_number, age):
        self.customers = {}
        self.__salary = salary
        self.__account_number = account_number
        self.__age = age
        self.__name = name
    

    def add_customer(self, name):
        self.customers[name] = {
            'salary': self.__salary,
            'account_number': self.__account_number,
            'age': self.__age
        }

    def get_customer_info(self, name):
        return self.customers.get(name, "Customer not found")

    def add_amount(self, account_number, amount):
        for customer, info in self.customers.items():
            if info['account_number'] == account_number:
                info['salary'] += amount
                return f"Amount added. New balance: {info['salary']}"
        return "Account number not found"

    def withdraw_amount(self, account_number, amount):
        for customer, info in self.customers.items():
            if info['account_number'] == account_number:
                if info['salary'] >= amount:
                    info['salary'] -= amount
                    return f"Amount withdrawn. New balance: {info['salary']}"
                else:
                    return "Insufficient balance"
        return "Account number not found"


Bank1 = Bank(5000, "Alice", "123456789", 30)
Bank1.add_customer("Alice")

print(Bank1.get_customer_info("Alice"))
```
