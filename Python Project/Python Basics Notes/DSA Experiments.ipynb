# merge sort / divide and conquer algorithm


```python
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2  # Finding the mid of the array
        L = arr[:mid]        # Dividing the elements into 2 halves
        R = arr[mid:]

        merge_sort(L)        # Sorting the first half
        merge_sort(R)        # Sorting the second half

        i = j = k = 0

        # Copy data to temp arrays L[] and R[]
        while i < len(L) and j < len(R):
            if L[i] < R[j]:
                arr[k] = L[i]
                i += 1
            else:
                arr[k] = R[j]
                j += 1
            k += 1

        # Checking if any element was left
        while i < len(L):
            arr[k] = L[i]
            i += 1
            k += 1

        while j < len(R):
            arr[k] = R[j]
            j += 1
            k += 1

# Example usage


data = [38, 27, 43, 3, 9, 82,6 ,7, 10]
print("Unsorted array:", data)
merge_sort(data)
print("Sorted array:", data)
```

```python
class Solution:
    def countPrimes(self, n: int) -> int:
        if n < 2:
            return 0

        my_numb = [True] * n
        my_numb[0] = my_numb[1] = False

        p = 2
        while p * p < n:
            if my_numb[p]:
                for i in range(p*p, n, p):
                    my_numb[i] = False
            p += 1

        return sum(my_numb)
```

# 72 sort colors


```python
class Solution:
    def sortColors(self, nums: list[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        low = 0
        mid = 0
        high = len(nums) - 1

        while mid <= high:
            if nums[mid] == 0:
                nums[low], nums[mid] = nums[mid], nums[low]
                low += 1
                mid += 1
            elif nums[mid] == 1:
                mid += 1
            else:  # nums[mid] == 2
                nums[mid], nums[high] = nums[high], nums[mid]
                high -= 1
```

# arranging coins


```python
class Solution:
    def arrangeCoins(self, n: int) -> int:
        left, right = 0, n

        while left <= right:
            mid = left + (right - left) // 2
            curr = mid * (mid + 1) // 2

            if curr == n:
                return mid
            if n < curr:
                right = mid - 1
            else:
                left = mid + 1

        return right
```

```python
    #135 - Candy

    class Solution:
        def candy(self, ratings: list[int]) -> int:
            n = len(ratings)
            candies = [1] * n

            # Left to right pass
            for i in range(1, n):
                if ratings[i] > ratings[i - 1]:
                    candies[i] = candies[i - 1] + 1

            # Right to left pass
            for i in range(n - 2, -1, -1):
                if ratings[i] > ratings[i + 1]:
                    candies[i] = max(candies[i], candies[i + 1] + 1)

            return sum(candies)
```

```python
# 678. Valid Parentheses String

class Solution:
    def checkValidString(self, s: str) -> bool:
        low = 0
        high = 0

        for char in s:
            if char == '(':
                low += 1
                high += 1
            elif char == ')':
                low -= 1
                high -= 1
            else:  # char == '*'
                low -= 1
                high += 1

            if high < 0:
                return False
            if low < 0:
                low = 0

        return low == 0
    

```

```python
# 242. Valid Anagram

class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False

        char_count = {}

        for char in s:
            char_count[char] = char_count.get(char, 0) + 1

        for char in t:
            if char not in char_count or char_count[char] == 0:
                return False
            char_count[char] -= 1

        return True
```

```python
# 153. Find Minimum in Rotated Sorted Array I

class Solution:
    def findMin(self, nums: list[int]) -> int:
        left, right = 0, len(nums) - 1

        while left < right:
            mid = left + (right - left) // 2

            if nums[mid] > nums[right]:
                left = mid + 1
            else:
                right = mid

        return nums[left]
```

```python
# 169. Majority Element

class Solution:
    def majorityElement(self, nums: list[int]) -> int:
        count = 0
        candidate = None

        for num in nums:
            if count == 0:
                candidate = num
            count += (1 if num == candidate else -1)

        return candidate
```

```python
#13. Roman to Integer

class Solution:
    def romanToInt(self, s: str) -> int:
        roman_numerals = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }

        total = 0
        prev_value = 0

        for char in reversed(s):
            value = roman_numerals[char]
            if value < prev_value:
                total -= value
            else:
                total += value
            prev_value = value

        return total    
```

```python
# 128. Longest Consecutive Sequence with array sorting

class Solution:
    def longestConsecutive(self, nums: list[int]) -> int:
        if not nums:
            return 0

        nums = list(set(nums))
        nums.sort()

        longest_streak = 1
        current_streak = 1

        for i in range(1, len(nums)):
            if nums[i] == nums[i - 1] + 1:
                current_streak += 1
            else:
                longest_streak = max(longest_streak, current_streak)
                current_streak = 1

        return max(longest_streak, current_streak)
```

```python
# 7. Reverse Integer

class Solution:

    def reverse(self, x: int) -> int:
        sign = -1 if x < 0 else 1
        x_abs_str = str(abs(x))
        reversed_str = x_abs_str[::-1]
        reversed_int = sign * int(reversed_str)

        if reversed_int < -2**31 or reversed_int > 2**31 - 1:
            return 0

        return reversed_int
```

```python
# 509. Fibonacci Number

class Solution:
    def fib(self, n: int) -> int:
        if n <= 1:
            return n

        a, b = 0, 1
        for _ in range(2, n + 1):
            a, b = b, a + b
        return b
```

```python
#164. Maximum Gap

class Solution:
    def maximumGap(self, nums: list[int]) -> int:
        if len(nums) < 2:
            return 0

        nums.sort()
        max_gap = 0

        for i in range(1, len(nums)):
            max_gap = max(max_gap, nums[i] - nums[i - 1])

        return max_gap
```

```python
# jump game 2

class Solution:
    def jump(self, nums: list[int]) -> int:
        jumps = 0
        current_end = 0
        farthest = 0

        for i in range(len(nums) - 1):
            farthest = max(farthest, i + nums[i])

            if i == current_end:
                jumps += 1
                current_end = farthest

        return jumps
```

```python
# candy

class Solution:
    def candy(self, ratings: list[int]) -> int:
        n = len(ratings)
        candies = [1] * n

        # Left to right pass
        for i in range(1, n):
            if ratings[i] > ratings[i - 1]:
                candies[i] = candies[i - 1] + 1

        # Right to left pass
        for i in range(n - 2, -1, -1):
            if ratings[i] > ratings[i + 1]:
                candies[i] = max(candies[i], candies[i + 1] + 1)

        return sum(candies)
```

```python
# 3074. Apple Redistribution into Boxes

class Solution:
    def minimumRedistribution(self, apples: list[int], boxes: int) -> int:
        total_apples = sum(apples)
        if total_apples % boxes != 0:
            return -1

        target = total_apples // boxes
        moves = 0

        for apple_count in apples:
            if apple_count > target:
                moves += apple_count - target

        return moves
```

```python
# 3074. Apple Redistribution into Boxes (LeetCode version)
# Given two lists: apple (apples in each box) and capacity (capacity of each box),
# Find the minimum number of boxes needed to store all apples.
# Each box can be used at most once, and you must use as few boxes as possible.

from typing import List

class Solution:
    def minimumBoxes(self, apple: List[int], capacity: List[int]) -> int:
        total_apples = sum(apple)
        capacity.sort(reverse=True)
        count = 0
        for cap in capacity:
            total_apples -= cap
            count += 1
            if total_apples <= 0:
                return count
        return -1  # Not enough capacity

# Example usage:
apple = [1, 3, 2]
capacity = [4, 3, 1, 2]
print(Solution().minimumBoxes(apple, capacity))  # Output: 2

apple2 = [5, 5, 5]
capacity2 = [4, 4, 4]
print(Solution().minimumBoxes(apple2, capacity2))  # Output: -1
```

```python
# LeetCode 3074. Apple Redistribution into Boxes
# Given a list of apples in each box and the number of boxes,
# find the minimum number of moves to redistribute apples so that all boxes have the same number.
# In one move, you can move one apple from one box to another.
# If impossible, return -1.

def minimum_moves(apples: list[int], boxes: int) -> int:
    total_apples = sum(apples)
    if total_apples % boxes != 0:
        return -1  # Impossible to distribute equally
    target = total_apples // boxes
    moves = 0
    for apple_count in apples:
        if apple_count > target:
            moves += apple_count - target
    return moves

# Example usage:
apples1 = [1, 0, 5]
boxes1 = 3
print(minimum_moves(apples1, boxes1))  # Output: 3

apples2 = [0, 2, 0]
boxes2 = 3
print(minimum_moves(apples2, boxes2))  # Output: -1
```
